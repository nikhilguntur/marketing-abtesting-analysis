```python
import pandas as pd
import numpy as np
from scipy import stats

# Load data
df = pd.read_csv('marketing_AB.csv')

# Basic info
print("Total users:", len(df))
print("Groups:", df['test group'].unique())
print()

# Separate groups
control = df[df['test group'] == 'psa']
treatment = df[df['test group'] == 'ad']

print(f"Control (PSA): {len(control)} users")
print(f"Treatment (Ad): {len(treatment)} users")
print()

# Calculate conversion rates
control_rate = (control['converted'].sum() / len(control)) * 100
treatment_rate = (treatment['converted'].sum() / len(treatment)) * 100
lift = treatment_rate - control_rate

print(f"Control conversion rate: {control_rate:.2f}%")
print(f"Treatment conversion rate: {treatment_rate:.2f}%")
print(f"Lift: {lift:+.2f} percentage points")
print()

# Z-test
control_conversions = control['converted'].sum()
treatment_conversions = treatment['converted'].sum()
total_conversions = control_conversions + treatment_conversions
total_users = len(df)

p_pool = total_conversions / total_users
se = np.sqrt(p_pool * (1 - p_pool) * (1/len(control) + 1/len(treatment)))
z = (treatment_rate/100 - control_rate/100) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"Z-statistic: {z:.4f}")
print(f"P-value: {p_value:.10f}")
print()

if p_value < 0.05:
    print("✅ STATISTICALLY SIGNIFICANT")
else:
    print("❌ NOT SIGNIFICANT")

# Recommendation
print()
if p_value < 0.05 and lift > 0.5:
    print("🎯 RECOMMENDATION: ✅ APPROVE campaign")
elif p_value < 0.05:
    print("🎯 RECOMMENDATION: ⚠️ MARGINAL - Consider cost vs lift")
else:
    print("🎯 RECOMMENDATION: ❌ REJECT campaign")
```
